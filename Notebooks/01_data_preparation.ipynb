# Cyclistic Bike-Share Data Preparation & Ingestion

## Objective
This notebook initiates the Cyclistic Bike-Share Analytics project by performing the critical steps of raw data ingestion and initial preparation. Key objectives include:
-   **Efficient Data Loading**: Consolidate 12 months of historical trip data from individual CSV files into a single, unified Pandas DataFrame.
-   **Schema & Type Enforcement**: Validate and convert data types for critical columns (e.g., `started_at`, `ended_at` to datetime, categorical features) to ensure data quality and optimize memory usage.
-   **Initial Data Quality Filtering**: Apply preliminary filters to address any immediate data inconsistencies or invalid records at the ingestion stage, ensuring a clean foundation for subsequent analysis.

## Data Lineage & Security
Maintaining data integrity, traceability, and adherence to data warehousing principles is paramount. This notebook implements the following:
-   **Zero Column Loss Principle**: All original columns from the raw source files are preserved during the initial loading phase. No data is prematurely dropped or aggregated, ensuring full fidelity to the source.
-   **Immutable Source Reference**: The raw CSV files are treated as the definitive, immutable source. This notebook reads data without modifying the original files.
-   **Traceability Mechanism**: A `source_file` column is temporarily introduced during the concatenation process to explicitly track the origin of each record. This facilitates auditing and debugging, and is subsequently removed once its lineage purpose is served.
-   **Optimized Persistence**: The prepared dataset is saved in an optimized format (e.g., Parquet) for efficient storage and high-performance access in downstream analytical processes.

The dataset consists of Cyclistic’s historical trip data for the last 12 months, stored as 12 CSV files, one for each month. The data has been made available by Motivate International Inc. under this license. Each file contains 13 fields:

- `ride_id`: Unique ID for each ride taken by the customer
- `rideable_type`: Type of bicycle among classic_bike, dock_bike and electric_bike
- `started_at`: Time at which the ride started
- `ended_at`: Time at which the ride ended
- `start_station_name`: Station at which the ride started
- `start_station_id`: The ID of the station at which the ride started
- `end_station_name`: Station at which the ride ended
- `end_station_id`: The ID of the station at which the ride ended
- `start_lat`: The latitude of the station at which the ride started
- `start_lng`: The longitude of the station at which the ride started
- `end_lat`: The latitude of the station at which the ride ended
- `end_lng`: The longitude of the station at which the ride ended
- `member_casual`: Type of user among member or casual

```python
#importing lib's
import os
from datetime import datetime
import numpy as np 
import pandas as pd
import matplotlib as plt
from pathlib import Path
```

```python
DATA_PATH = Path("D:\cicyle project\Cyclistic-dataset") 
csv_files = sorted(DATA_PATH.glob("*.csv"))

csv_files
```

```python
df_list = []

for file in csv_files:
    temp_df = pd.read_csv(file)
    temp_df["source_file"] = file.name
    df_list.append(temp_df)

df = pd.concat(df_list, ignore_index=True)
```

```python
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])
```

```python
categorical_cols = ['rideable_type', 'start_station_name', 'end_station_name', 'member_casual', 'source_file']
for col in categorical_cols:
    df[col] = df[col].astype('category')
```

```python
df
```

```python
df = df.drop(columns=['source_file'])
```

```python
df.to_parquet('date_prepared', index= False)
```
