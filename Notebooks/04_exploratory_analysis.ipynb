# Exploratory Behavioral Analysis

This notebook explores behavioral differences between annual members and casual riders using engineered features.

The objective is to identify:
- riding behavior patterns
- commuting versus leisure usage
- temporal demand differences
- potential membership conversion opportunities

The analysis focuses on generating business-relevant insights rather than descriptive statistics alone.

### Business Questions Addressed

This notebook serves as a critical step in translating raw bike-share data into actionable business intelligence. By leveraging engineered temporal and behavioral features, we aim to answer key questions that directly impact Cyclistic's strategic objectives, particularly in driving annual membership growth. The primary business questions addressed are:

1.  **Rider Behavior Patterns:** How do the riding habits of annual members differ from those of casual riders across various dimensions (e.g., ride duration, frequency, route choices)?
2.  **Commuting vs. Leisure Usage:** What are the distinct patterns that indicate whether riders are using Cyclistic bikes for daily commuting versus recreational or leisure purposes?
3.  **Temporal Demand Differences:** Are there significant variations in bike usage between member types based on time of day, day of week, or season, and what do these differences imply for service optimization and marketing?
4.  **Membership Conversion Opportunities:** Can we identify specific casual rider behaviors or usage contexts that present the highest potential for conversion to annual memberships?
5.  **High-Traffic Corridors:** Which routes or station pairs are most frequently used by both casual and member riders, and how can this information be leveraged for targeted promotions or infrastructure improvements?

```python
#Ride Length Distribution
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="ride_length_min",
    hue="member_casual",
    bins=50,
    kde=True
)

plt.xlabel("Ride Length (Minutes)")
plt.ylabel("Ride Count")

plt.show()
```

```python
df['member_casual'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Member vs Casual Split')
plt.show()
```

```python
avg = df.groupby('member_casual')['ride_length_min'].mean()

avg.plot(kind='bar')

plt.title('Average Ride Length by User Type')
plt.ylabel('Minutes')
plt.show()
```

```python
is_weekend = (
    df.groupby(['member_casual', 'is_weekend'])
    .size()
    .reset_index(name='ride_count')
)

sns.barplot(
    data=is_weekend,
    x="is_weekend",
    y="ride_count",
    hue="member_casual"
)

plt.title("day type prefernce")
plt.xlabel("Day Type")
plt.ylabel("Ride Count")

plt.show()
```

```python
#Hourly Usage Trends
hour = df.groupby(['hour','member_casual']).size().unstack()

hour.plot(kind='line', figsize=(12,6))

plt.title('Hourly Usage Trends')
plt.ylabel('Ride Count')
plt.xticks(range(0, 24))

plt.show()
```

```python
#Monthly Usage Trends
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_usage['month'] = pd.Categorical(
    monthly_usage['month'],
    categories=month_order,
    ordered=True
)

monthly_usage = monthly_usage.sort_values('month')

plt.figure(figsize=(14,6))

sns.lineplot(
    data=monthly_usage,
    x='month',
    y='rides',
    hue='member_casual',
    marker='o'
)

plt.title('Monthly Ride Volume by Rider Type')
plt.xlabel('Month')
plt.ylabel('Number of Rides')

plt.xticks(rotation=45)
plt.show()
```

```python
#Bike type usage
bike = df.groupby(['member_casual', 'rideable_type']).size().unstack()
bike.plot(kind='bar')
plt.title('Bike Type by User Type')
plt.show()
```
